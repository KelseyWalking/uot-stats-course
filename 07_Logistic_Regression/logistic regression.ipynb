```python
''' we want a transformation that makes the range of possibilities on the left hand side
of the equation equal to the range of possibilities for the right hand side; if there was no transformation for this equation, the left hand side could only take values between 0 and 1, but the right
hand side could take values outside of this range. A common transformation for pi
is the logit
transformation'''
'''
MEANING: we find  positive (to infinity) or negative (to infinity) value for each possible p (0<p<1) 
we took these values (a fixed line) and map ax1+bx2+... to this line. 
'''


import numpy as np
import matplotlib.pyplot as plt

# Define the logit function
def logit(p):
    return np.log(p / (1 - p))

# Generate a range of p values between 0 and 1, but not including 0 or 1 to avoid division by zero
p_values = np.linspace(0.01, 0.99, 500)

# Calculate the corresponding logit values
logit_values = logit(p_values)

# Plot the logit function
plt.figure(figsize=(10, 6))
plt.plot(p_values, logit_values, label=r'$\log_e\left(\frac{p}{1 - p}\right)$')
plt.xlabel(r'$p$', fontsize=14)
plt.ylabel(r'$\text{logit}(p)$', fontsize=14)
plt.title('Logit Function', fontsize=16)
plt.grid(True)
plt.legend(fontsize=14)
plt.show()
```
